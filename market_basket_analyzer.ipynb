```python
import numpy as np
import pandas as pd
import csv
```

```python
transactions = []
with open('Market_Basket_Optimisation.csv', 'r') as file:
    reader = csv.reader(file)
    for row in reader:
        # Drop any empty strings if they exist
        clean_row = [item for item in row if item != '']
        transactions.append(clean_row)

print(f"Total transactions: {len(transactions)}")
for i in range(5):
    print(f"Receipt {i+1}: {transactions[i]}")
```

```python
from mlxtend.preprocessing import TransactionEncoder
```

```python
te = TransactionEncoder()
```

```python
te_array = te.fit_transform(transactions)
```

```python
df = pd.DataFrame(te_array, columns = te.columns_)
```

```python
df.shape
```

```python
from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules
```

```python
frequent_items = apriori(df, min_support =0.01, use_colnames=True)
```

```python
rules = association_rules(frequent_items, metric='confidence', min_threshold=0.2)
```

```python
rules = rules.sort_values(by='lift', ascending=False)
```
